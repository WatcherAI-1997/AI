## ✅ Шаг 2: Проверка файлов

Убедимся что все файлы на месте

```
# ═══════════════════════════════════════════════════════════════
# Проверка наличия файлов
# ═══════════════════════════════════════════════════════════════

import os

required_files = [
    'tamerlane_model.py',
    'turkic_knowledge_base.py',
    'universal_knowledge.py',
    'advanced_knowledge.py',
    'world_languages.py',
    'ancient_languages.py',
    'tengri_knowledge.py'
]

print("🔍 Проверка файлов...\n")

missing = []
for file in required_files:
    if os.path.exists(file):
        size = os.path.getsize(file) / 1024
        print(f"✅ {file:30} ({size:6.1f} KB)")
    else:
        print(f"❌ {file:30} ОТСУТСТВУЕТ!")
        missing.append(file)

if missing:
    print(f"\n⚠️  Отсутствует {len(missing)} файлов!")
    print("Проверьте что репозиторий создан правильно.")
else:
    print("\n✅ Все файлы на месте!")
```

## 🧪 Шаг 4: Тестирование

Функция для удобного тестирования модели

```
# ═══════════════════════════════════════════════════════════════
# Функция тестирования
# ═══════════════════════════════════════════════════════════════

def test(question):
    """Тестирование модели"""
    print(f"\n{'='*70}")
    print(f"❓ ВОПРОС: {question}")
    print('='*70)
    
    # Демо-ответы для основных тем
    demo_responses = {
        'тенгри': '''**ТЕНГРИ (𐱅𐰭𐰼𐰃) - ВЕРХОВНЫЙ БОГ ТЮРКОВ**

Тенгри - верховное божество в древней тюркской религии.

🌌 **Атрибуты:**
• Творец неба и земли
• Всемогущий и всевидящий  
• Справедливый судья

🏹 **В истории:**
Все тюркские каганы правили "по воле Тенгри".
На Орхонских надписях: 𐱅𐰭𐰼𐰃 = ТЕНГРИ

Тәңірі жарылқасын! (Да благословит Тенгри!)''',
        
        'умай': '''**УМАЙ (ҰМАЙ) - БОГИНЯ-МАТЬ**

Умай - богиня плодородия и покровительница женщин и детей.

👶 **Роль:**
• Защитница детей и рожениц
• Богиня плодородия
• Покровительница семейного очага

🌸 **Символы:**
• Нить жизни
• Золотая колыбель''',
        
        'python': '''**PYTHON - ЯЗЫК ПРОГРАММИРОВАНИЯ**

Python - самый популярный язык для AI!

🚀 **Для AI:**
• TensorFlow, PyTorch - нейросети
• scikit-learn - машинное обучение
• Pandas, NumPy - анализ данных

💡 **Преимущества:**
• Простой синтаксис
• Огромная экосистема
• Активное сообщество''',
        
        'орхон': '''**ОРХОНСКИЙ АЛФАВИТ**

Орхоно-енисейская руница - древнетюркское письмо VI-X веков.

📜 **Основные руны:**
𐰀=a, 𐰁=ä, 𐰂=b, 𐰃=i, 𐰴=q, 𐰼=r, 𐱅=t, 𐰭=ng

🌌 **Священное слово:**
𐱅𐰭𐰼𐰃 = ТЕНГРИ (Бог Неба)

🏛️ **Знаменитые надписи:**
• Кюль-Тегинская надпись (732 г.)
• Надпись Бильге-кагана (735 г.)'''
    }
    
    # Определяем ответ
    q_lower = question.lower()
    response = None
    
    for key, ans in demo_responses.items():
        if key in q_lower:
            response = ans
            break
    
    if not response:
        response = f"Демо-ответ на: {question}\n\n[Подключите Claude/Gemini API для полных AI-ответов]"
    
    # Генерация с RAG
    messages = [{"role": "user", "content": question}]
    result = model.generate(
        messages=messages,
        base_model_fn=lambda m, **k: response
    )
    
    # Метаданные
    print(f"\n📊 RAG Обогащение: {'✅ ДА' if result['enhanced'] else '❌ НЕТ'}")
    
    if result['enhanced']:
        print(f"   • Контекст: {result['metadata']['context_length']} символов")
        print(f"   • Релевантность: {result['metadata']['max_relevance']:.2f}")
        if result['metadata']['categories']:
            cats = ', '.join(result['metadata']['categories'])
            print(f"   • Категории: {cats}")
    
    # Ответ
    print(f"\n💬 Ответ:\n\n{result['response']}\n")

print("✅ Функция test() готова!")
print("\n📝 Используйте: test('ваш вопрос')")
```
