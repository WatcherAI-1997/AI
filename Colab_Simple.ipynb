## Шаг 3: Тестируйте!

Запустите примеры ниже:

```
# Функция для тестирования
def test(question):
    print(f"\n{'='*70}")
    print(f"❓ {question}")
    print('='*70)
    
    # Простые демо-ответы
    answers = {
        'тенгри': '''**ТЕНГРИ (𐱅𐰭𐰼𐰃) - БОГ НЕБА**
        
Верховное божество тюрков.
• Творец неба и земли
• Всемогущий и всевидящий
• Справедливый судья

На орхонских надписях: 𐱅𐰭𐰼𐰃 = ТЕНГРИ''',
        
        'умай': '''**УМАЙ - БОГИНЯ-МАТЬ**
        
Покровительница женщин и детей.
• Защитница рожениц
• Богиня плодородия
• Хранительница очага''',
        
        'python': '''**PYTHON**
        
Популярный язык для AI!
• TensorFlow, PyTorch
• scikit-learn
• Pandas, NumPy'''
    }
    
    # Определяем ответ
    q_lower = question.lower()
    response = "Демо-ответ. Подключите Claude/Gemini API для полных ответов."
    
    for key, ans in answers.items():
        if key in q_lower:
            response = ans
            break
    
    # Генерация с RAG
    messages = [{"role": "user", "content": question}]
    result = model.generate(
        messages=messages,
        base_model_fn=lambda m, **k: response
    )
    
    # Результат
    print(f"\n📊 RAG: {'✅' if result['enhanced'] else '❌'}")
    if result['enhanced']:
        print(f"   Контекст: {result['metadata']['context_length']} символов")
    
    print(f"\n💬 Ответ:\n{result['response']}\n")

print("✅ Функция test() готова!")
print("\n📝 Используйте: test('ваш вопрос')")
```

### Примеры:

```
# Пример 1
test("Кто такой Тенгри?")
```

```
# Пример 2
test("Расскажи о богине Умай")
```

```
# Пример 3
test("Что означает 𐱅𐰭𐰼𐰃?")
```

```
# Пример 4
test("Что такое Python?")
```

```
# Свой вопрос
test("ВАШ ВОПРОС ЗДЕСЬ")
```
