## 🧪 Шаг 4: Тестирование модели

```python
# ═══════════════════════════════════════════════════════════════
# 🧪 Функция для тестирования
# ═══════════════════════════════════════════════════════════════

def test_query(question):
    """Тестирование вопроса"""
    print(f"\n{'='*80}")
    print(f"❓ ВОПРОС: {question}")
    print('='*80)

    messages = [{"role": "user", "content": question}]

    # Мок-функция для демонстрации (пока без реального AI)
    def mock_model(msgs, **kwargs):
        user_msg = msgs[-1]['content']
        
        # Простые ответы на ключевые слова
        if 'тенгри' in user_msg.lower() or 'tengri' in user_msg.lower() or '𐱅𐰭𐰼𐰃' in user_msg:
            return """**ТЕНГРИ (𐱅𐰭𐰼𐰃) - ВЕРХОВНЫЙ БОГ ТЮРКОВ**

Тенгри - это верховное божество в древней тюркской религии.

🌌 **Атрибуты:**
• Творец неба и земли
• Всемогущий и всевидящий
• Справедливый судья

🏹 **В истории:**
Все тюркские каганы правили "по воле Тенгри".
На Орхонских надписях: 𐱅𐰭𐰼𐰃 = ТЕНГРИ

Тәңірі жарылқасын!"""
        
        elif 'умай' in user_msg.lower():
            return """**УМАЙ - БОГИНЯ-МАТЬ**

Умай - богиня плодородия и покровительница женщин и детей.

👶 **Роль:**
• Защитница детей и рожениц
• Богиня плодородия
• Покровительница семейного очага"""
        
        elif 'python' in user_msg.lower():
            return """**PYTHON - ЯЗЫК ПРОГРАММИРОВАНИЯ**

Python - самый популярный язык для AI!

🚀 **Для AI:**
• TensorFlow, PyTorch - нейросети
• scikit-learn - машинное обучение
• Pandas, NumPy - анализ данных"""
        
        else:
            return f"Ответ на: {user_msg}\n\n[Для полных ответов подключите Claude или Gemini API]"

    # Генерация ответа
    result = model.generate(
        messages=messages,
        base_model_fn=mock_model,
        temperature=0.7
    )

    # Показать метаданные
    print(f"\n📊 Метаданные:")
    print(f"   • RAG обогащение: {'✅ ДА' if result['enhanced'] else '❌ НЕТ'}")
    
    if result['enhanced']:
        print(f"   • Контекст: {result['metadata']['context_length']} символов")
        print(f"   • Релевантность: {result['metadata']['max_relevance']:.2f}")
        if result['metadata']['categories']:
            cats = ', '.join(result['metadata']['categories'])
            print(f"   • Категории: {cats}")

    # Показать ответ
    print(f"\n💬 Ответ:\n")
    print(result['response'])
    print()

print("✅ Функция test_query() готова!")
print("\n📝 Используйте: test_query('ваш вопрос')")
```

## 🔑 Опционально: Подключение реального AI

Для получения полноценных ответов подключите Claude или Gemini API:

```python
# ═══════════════════════════════════════════════════════════════
# 🔑 ПОДКЛЮЧЕНИЕ CLAUDE API
# ═══════════════════════════════════════════════════════════════

from anthropic import Anthropic

# ⚠️ ВСТАВЬТЕ ВАШ API КЛЮЧ
ANTHROPIC_API_KEY = "sk-ant-ваш-ключ-здесь"  # Замените на ваш ключ

client = Anthropic(api_key=ANTHROPIC_API_KEY)

def claude_model(messages, **kwargs):
    """Реальная модель Claude"""
    response = client.messages.create(
        model="claude-3-5-sonnet-20241022",
        max_tokens=2000,
        messages=messages
    )
    return response.content[0].text

print("✅ Claude API подключен!")
print("\n📝 Теперь используйте:")
print("   test_query_real('ваш вопрос')")

# Новая функция с реальным AI
def test_query_real(question):
    """Тестирование с реальным Claude"""
    print(f"\n{'='*80}")
    print(f"❓ ВОПРОС: {question}")
    print('='*80)

    messages = [{"role": "user", "content": question}]

    result = model.generate(
        messages=messages,
        base_model_fn=claude_model,  # Используем реальный Claude
        temperature=0.7
    )

    print(f"\n📊 RAG: {'✅' if result['enhanced'] else '❌'}")
    if result['enhanced']:
        print(f"   • Контекст: {result['metadata']['context_length']} символов")

    print(f"\n💬 Ответ:\n")
    print(result['response'])
    print()
```

```python
# ═══════════════════════════════════════════════════════════════
# 🔑 ПОДКЛЮЧЕНИЕ GEMINI API (альтернатива)
# ═══════════════════════════════════════════════════════════════

import google.generativeai as genai

# ⚠️ ВСТАВЬТЕ ВАШ API КЛЮЧ
GOOGLE_API_KEY = "ваш-ключ-здесь"  # Замените на ваш ключ

genai.configure(api_key=GOOGLE_API_KEY)
gemini_model = genai.GenerativeModel('gemini-1.5-pro')

def gemini_fn(messages, **kwargs):
    """Реальная модель Gemini"""
    prompt = messages[-1]['content']
    response = gemini_model.generate_content(prompt)
    return response.text

print("✅ Gemini API подключен!")
print("\n📝 Используйте gemini_fn вместо claude_model")
```
