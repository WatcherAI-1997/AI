## 💬 Шаг 7: Интерактивная генерация

Попробуйте свои промпты!

```python
# Загрузка модели для интерактивного использования
import torch
from model.tokenizer import TamerlaneTokenizer
from model.architecture import TamerlaneGPT

print("⏳ Загрузка модели...")

device = 'cuda' if torch.cuda.is_available() else 'cpu'

# Загрузка токенизатора
tokenizer = TamerlaneTokenizer()
tokenizer.load('model/tokenizer.json')

# Загрузка модели
checkpoint = torch.load('checkpoints/best_model.pt', map_location=device)
model_info = checkpoint['model_info']

model = TamerlaneGPT(
    vocab_size=tokenizer.get_vocab_size(),
    d_model=model_info['d_model'],
    num_layers=model_info['num_layers']
)
model.load_state_dict(checkpoint['model_state_dict'])
model.to(device)
model.eval()

print("✅ Модель загружена!\n")

# Функция генерации
def generate_text(prompt, max_length=100, temperature=0.8):
    # Токенизация
    input_ids = [tokenizer.special_tokens['<BOS>']] + tokenizer.encode(prompt)
    input_tensor = torch.tensor([input_ids], dtype=torch.long).to(device)
    
    # Генерация
    with torch.no_grad():
        output = model.generate(input_tensor, max_new_tokens=max_length, temperature=temperature, top_k=50)
    
    # Декодирование
    output_ids = output[0].tolist()
    filtered_ids = [id for id in output_ids if id not in [0, 2, 3]]  # Убираем PAD, BOS, EOS
    
    return tokenizer.decode(filtered_ids)

print("🎯 Попробуйте свои промпты:\n")

# Примеры
test_prompts = [
    "Тенгри",
    "Тәмірлан",
    "Математика"
]

for prompt in test_prompts:
    print(f"\n📝 Промпт: {prompt}")
    print("-" * 80)
    result = generate_text(prompt, max_length=50)
    print(result)
    print("-" * 80)
```
